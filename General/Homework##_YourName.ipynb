```python
#Part A
#Creating a function that rolls two dice
def rolldice():
    #Rolling the first die
    value1 = r.randint(1,6)
    #Rolling the second die
    value2 = r.randint(1,6)
    return value1, value2

#Rolling the dice
value1, value2 = rolldice()

#Printing what was rolled
print(value1, value2)
```

#### Explanation: 
All that was done was I created a function that doesn't take any values. Then it finds two values which are a random integer from 1 to 6. Finally, it returns those two values, and prints them. 

```python
#Double six index
doublesix = 0

#Rolls the two die 200 times
for i in range(1000000):
    #Rolls the dice
    value1, value2 = rolldice()
    #Checks to see if they are both sixes
    if value1 == value2 == 6:
        doublesix+=1

print(doublesix)
#Prints 1/36 compared to doublesix/1000000
print(1/36, doublesix/1000000)

```
