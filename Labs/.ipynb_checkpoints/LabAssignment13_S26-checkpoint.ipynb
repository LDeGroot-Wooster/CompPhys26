## Newton's Method - single variable (22 pts)

Solve the same problem from class used for binary search, but analyzed using Newton's Method instead. 
Find the roots of $$ f(x) = e^{x-\sqrt{x}}-x$$

To do this: 

1) Take the derivative by hand. You can use something like WolframAlpha to check your derivative. Write it in markdown. 

2) Write two functions - `f(x)` that returns the result of the function for a given $x$ value and `df(x)` one that returns the derivative. 

3) Make a plot of the function from $x=0$ to $x=5$. Use this to estimate the roots. 

4) Apply the functions to find both roots. 

*You must clearly comment your code and include a markdown cell explaining your process*

The derivative is: 

$$e^{x-\sqrt{x}}\left(1-\frac{1}{2\sqrt{x}}\right)-1 $$

$$\frac{d}{dx}\left(\text{e}^{x-\sqrt{x}}-x\right)=\frac{d}{dx}(x-\sqrt{x})\text{e}^{x-\sqrt{x}}-1=\left(1-\frac{1}{2\sqrt{x}}\right)\text{e}^{x-\sqrt{x}}-1$$

```python
# part 2
# define a function of the polynomial
def f(x):
    return (math.e**(x-np.sqrt(x)))-x

# define function that is the derivative of the polynomial 
def df(x):
    return ((math.e**(x-np.sqrt(x)))*(1-(1/(2*np.sqrt(x)))))-1
```

```python
# part 3
#a range of x values
x_vals = np.linspace(0,5,100)
y_vals = np.zeros(len(x_vals))

for i in range(len(x_vals)): #use function f(x) to find y values for each x value
    y_vals[i] = f(x_vals[i])

#plot
plt.scatter(x_vals, y_vals, s = 10)
plt.grid()
```

```python
# part 4
guesses = np.array([1, 2.5]) #these look to be the approximate values of the roots

#desired accuracy and initial value of error to initiate while loop
desired_acc = 1e-10
max_err = 1.0

#performs Newton's method
while max_err>desired_acc:
    deltas = f(guesses)/df(guesses)
    guesses -= deltas
    max_err = max(np.abs(deltas))

print(guesses)
```

In step 2, I define f(x) and df(x) as the function given and its derivative. In step 3, I plotted the function for 100 values of x between 0 and 5 to estimate the roots. They look to be at about 1 and 2.5, so I entered those guesses into my guesses array. I set the desired accuracy at 1e-10. Since my max_err is greater than my desired accuracy, tghe while loop initiates. For each guess it calculates a delta (f(x)/df(x)) and then subtracts that delta from the guess. The max_err is the absolute value of the largest delta. The while loop repeats until that value is less than the desired accuracy.

## Newton's Method - many variables (22 pts)

Real circuits, however, often include nonlinear components.  To solve for the behavior of these circuits we need to solve nonlinear equations.

Consider the following simple circuit, a variation on the classic

We will use Newton's Method to solve for the voltages in the 
The resistors obey the normal Ohm law, but the diode obeys the diode equation:
$$ I = I_0 (e^{V/V_T} - 1),$$
where $V$ is the voltage across the diode and $I_0$ and $V_T$ are constants.

The Kirchhoff current law says that the total net current flowing into or out of every point in a circuit must be zero.  Applying the law to voltage $V_1$ in the circuit above we get
$$
{V_1-V_+\over R_1} + {V_1\over R_2}
  + I_0\bigl[e^{(V_1-V_2)/V_T} - 1\bigr] = 0.
$$

   
1) In a markdown cell, derive the corresponding equation for voltage $V_2$.

2) Solve the two nonlinear equations for the voltages $V_1$ and $V_2$ using Newton's method to solve the equations. This means you need to first take the derivatives of each equation with respect to each variable (so you'll have 4 equations). Define the Jacobian matrix and determine your desired accuracy. Setup the matrix necessary to analyze using the `np.linalg.solve()` function. Apply the following conditions: 

$$\begin{align*}
V_+ &= 5\,\textrm{V}, \\
R_1 &= 1\,\mathrm{k}\Omega, \qquad
R_2 = 4\,\mathrm{k}\Omega, \qquad
R_3 = 3\,\mathrm{k}\Omega, \qquad
R_4 = 2\,\mathrm{k}\Omega, \\
I_0 &= 3\,\textrm{nA},\qquad

\end{align*}$$

3) The electronic engineer's rule of thumb for diodes is that the voltage across a (forward biased) diode is always about 0.6 volts. Confirm that your results agree with this rule by taking the difference between your results and explain briefly. 

*You must clearly comment your code and include a markdown cell explaining your process*

Part 1:

Voltage in/out through resistor $R_3$: $$\frac{V_2-V_{+}}{R_3}$$
Voltage in/out through resistor $R_4$: $$\frac{V_2-0}{R_4}$$
Voltage in/out through diode: $$-I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right)$$
Add the all together and set them equal to 0: $$\frac{V_2-V_{+}}{R_3} + \frac{V_2}{R_4} -I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right) = 0$$

Part 2:
$$\frac{\partial}{\partial V_1}\left(\frac{V_1-V_{+}}{R_1} + \frac{V_1}{R_2} + I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right)\right) = \frac{1}{R_1}+\frac{1}{R_2}+\frac{I_0}{V_T}\left(\text{e}^{(V_1 - V_2)/V_T}\right)$$
$$\frac{\partial}{\partial V_2}\left(\frac{V_1-V_{+}}{R_1} + \frac{V_1}{R_2} + I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right)\right) = -\frac{I_0}{V_T}\left(\text{e}^{(V_1 - V_2)/V_T}\right)$$
$$\frac{\partial}{\partial V_1}\left(\frac{V_2-V_{+}}{R_3} + \frac{V_2}{R_4} - I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right)\right) = -\frac{I_0}{V_T}\left(\text{e}^{(V_1 - V_2)/V_T}\right)$$
$$\frac{\partial}{\partial V_2}\left(\frac{V_2-V_{+}}{R_3} + \frac{V_2}{R_4} - I_0\left(\text{e}^{(V_1 - V_2)/V_T}-1\right)\right) = \frac{1}{R_3}+\frac{1}{R_4}+\frac{I_0}{V_T}\left(\text{e}^{(V_1 - V_2)/V_T}\right)$$

```python
# part 2
#define needed constants
V_plus = 5 #V
R_1 = 1000 #Ohm
I_0 = 3e-9 #Amp
R_2 = 4000 #Ohm
V_T = 0.05 #V
R_3 = 3000 #Ohm
R_4 = 2000 #Ohm

#define both functions
def f1(V_1,V_2):
    return ((V_1-V_plus)/R_1) + (V_1/R_2) + (I_0*((math.e**((V_1-V_2)/V_T))-1))

def f2(V_1,V_2):
    return ((V_2-V_plus)/R_3) + (V_2/R_4) - (I_0*((math.e**((V_1-V_2)/V_T))-1))

#guesses for voltages
V_1 = 1.0
V_2 = 1.0

#desired accuracy
desired_acc = 1e-10
err = np.sqrt(V_1**2+V_2**2)

#while loop perform newton's method
while err > desired_acc:
    #partial derivatives
    a = (1/R_1)+(1/R_2)+((I_0/V_T)*(math.e**((V_1-V_2)/V_T))) #df1/V_1
    b = -(I_0/V_T)*(math.e**((V_1-V_2)/V_T))                  #df1/V_2
    c = -(I_0/V_T)*(math.e**((V_1-V_2)/V_T))                 #df2/V_1
    d = (1/R_3)+(1/R_4)+((I_0/V_T)*(math.e**((V_1-V_2)/V_T))) #df2/V_2
    #Jacobian matrix
    A = np.array([[a,b],
                  [c,d]],float)
    v = np.array([f1(V_1,V_2),f2(V_1,V_2)], float)
    vals = np.linalg.solve(A, v)
    V_1 -= vals[0]
    V_2 -= vals[1]
    err = np.sqrt(vals[0]**2+vals[1]**2) #calculate error

print("V_1 =", V_1, "V_2 = ", V_2)
```

```python
# part 3
print("Difference =", round(V_1-V_2, 4), "Volts") #differnce ~ 0.6 V
```

As expected, the voltage difference across the diode is about 0.6V. To find this I first defined all constant and two equations f1 (the sum of voltages at $V_1$ = 0) and f2 (the sum of voltages at $V_2$ = 0). I then took the partial derivatives in terms of both $V_1$ and $V_2$ for both of these equations and plugged them in to the Jacobian matrix A. The solution matrix v contains the two original equations. I used the linalg.solve module to find the values to solve the matrix and then subtracted them from my initial guesses for $V_1$ and $V_2$. I recalculated the partials and repeated the process until the error was no longer greater than the desired accuracy.
