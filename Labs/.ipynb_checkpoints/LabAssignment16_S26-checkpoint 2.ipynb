### The Shooting Method

Find a solution to the boundary value problem: 
$$\frac{d^2y}{dt^2} = -y$$
where $y$ is the position of an object, and $y(t=0) = 1$ and $y(t=1) = 2$. To do this, complete the following steps: 

#### Step 1: In a markdown, write the two simultaneous single-order differential equations that you will use to solve this second-order differential equation. (Convert the cell below to markdown first)

$$\frac{dy}{dt}=v$$ and $$\frac{dv}{dt}=-y$$

#### Step 2: Write a function `f(r,t)` that will calculate the two equations written in step 1 using a vector `r` that containts two elements. 

```python
def f(r,t): #takes r (a vector)
    y = r[0]
    v = r[1]
    f_y = v #eq 1
    f_v = -y #eq 2
    return np.array([f_y,f_v], float) #returns a vector
```

#### Step 3: Define any constants, boundary conditions, and arrays necessary for the problem and solve for the initial velocity $\left(\frac{dy}{dt}\right)$. 

```python
#v0 = ?
y0 = 1
yf = 2
t0 = 0 #s
tf = 1 #s
N_steps = 1000
step = (tf-t0)/N_steps
t_arr = np.arange(t0, tf+step, step) #array of 1000 times between 0 and 1
desired_acc = 1e-10
```

#### Step 4: Use fourth-order Runge-Kutta combined with the binary search method to find the initial velocity ($\frac{dy}{dt}$) that solves the boundary value problem. Note: in this case the final point is not 0. In order to use binary search, you will need to subtract the final height from the runge-kutta found height using your initial guess to see if it is zero or not. 

```python
#apply RK4
def height(v_guess,t_arr,step):
    r = np.array([y0,v_guess], float)
    for t in t_arr:
        k1 = step*f(r,t)
        k2 = step*f(r+0.5*k1, t+0.5*step)
        k3 = step*f(r+0.5*k2, t+0.5*step)
        k4 = step*f(r+k3,t+step)
        r += (k1 + 2*k2 + 2*k3 + k4)/6
    return r[0] #returns velocity

#low guess high guess
v1 = 0.01
v2 = 500
y1 = height(v1, t_arr, step) - yf
y2 = height(v2, t_arr, step) - yf

if y1*y2 < 0: #check opposite signs
    while abs(y2-y1)>desired_acc:
        v_prime = (v1+v2)/2
        y_prime = height(v_prime, t_arr, step) - yf
        if y1*y_prime>0:   #same sign
            v1 = v_prime
            y1 = y_prime
        else:
            v2 = v_prime
            y2 = y_prime
        
v_init_f = (v1+v2)/2 #calculate final value
print("Initial velocity =", round(v_init_f,5), "m/s.")
```

#### Step 5: Analyze the accuracy of your result. One way you could do this is to plug your two now initial conditions back into fourth-order runge-kutta to solve for all $y$ values and make a plot of $y$ vs. $t$. In a markdown cell, briefly comment on the results. 

```python
#arrays to store values to plot
t_arr = np.arange(t0, tf+2*step, step)
y_arr = np.zeros(len(t_arr))
v_arr = np.zeros(len(t_arr))

r = np.array([y0,v_init_f], float) #put final v value calculated above in
for i in range(len(t_arr)):
    y_arr[i] = r[0]
    v_arr[i] = r[1]
    k1 = step*f(r,t_arr[i])
    k2 = step*f(r+0.5*k1, t_arr[i]+0.5*step)
    k3 = step*f(r+0.5*k2, t_arr[i]+0.5*step)
    k4 = step*f(r+k3,t_arr[i]+step)
    r += (k1 + 2*k2 + 2*k3 + k4)/6

print("The calculated end height is", y_arr[-1]) #should be about 2

#Plt velocity and height vs. time
plt.scatter(t_arr, y_arr, s=4, label = 'Height')
plt.scatter(t_arr, v_arr, s=4, label = "Velocity")
plt.legend()
plt.xlabel("Time")
plt.grid()
plt.show()
```

I use Runge Kutta and binary search similarly to the lab notebook to find the initial velocity, about 1.7 m/s. The only difference is that when I calculate y1, y2, and yprime, I subtract the yf value, 2 meters. I plugged my final value for the initial velocity back into RK4 and got a final height of almost exactly 2. Yay! The plot above shows the height and velocity vs. time.

### Eigenvalue Problems 

Find the smallest positive eigenvalue $z$ for the equation:
	    $$\frac{d^2y}{dt^2} = -z y$$
	    where $y(t =-1) = 0$ and $y(t = 1) = 0$. Try searching in the range $0.0 \leq z \leq 10.0$. To do this, complete the following steps:

#### Step 1: define the function `f(r,t,z)` that applies values (including your initial guess of $z$ and a vector `r` that containts two elements) to the two first-order differential equations needed for this problem. You may find it beneficial to write our your two first-order differential equations first. 

$$\frac{dy}{dt}=v$$ and $$\frac{dv}{dt}=-zy$$

```python
def f(r,t,z): #r is a vector
    y = r[0]
    v = r[1]
    f_y = v #eq 1
    f_v = -z*y #eq 2
    return np.array([f_y, f_v], float) #returns vector
```

#### Step 2: Define any constants, time constraints, and initial/boundary values necessay to apply fourth-order runge-kutta and either the binary search or secant method to solve this problem

```python
y0 = 0.0
yf = 0.0
t0 = -1
tf = 1
v0 = 0.01 #guess
N = 1000
step = (tf-t0)/N
t_arr = np.arange(t0, tf+step, step)
```

#### Step 3: Solve for $z$ using fourth-order runge-kutta and the root method of your choice (binary, secant)

```python
def eigen_RK4(z):
    r = np.array([y0, v0],float)
    for t in t_arr:
        k1 = step*f(r,t,z)
        k2 = step*f(r+0.5*k1, t+0.5*step, z)
        k3 = step*f(r+0.5*k2, t+0.5*step, z)
        k4 = step*f(r+k3, t+step, z)
        r += (k1 + 2*k2 + 2*k3 + k4)/6
    return r[0] #returns velocity
        
#secant method
z1 = 0.0
z2 = 10.0
y2 = eigen_RK4(z1)
desired_acc = 1e-10

while np.abs(z2-z1)>desired_acc:
    y1, y2 = y2, eigen_RK4(z2)
    z1, z2 = z2, z2 - (y2*(z1-z2)/(y1-y2))

print("z =", z2)
```

#### Step 4: Does your value of $z$ make sense? Show below by plugging initial conditions back into fourth-order Runge-Kutta and plotting $y$ vs. $t$. What happens if you change the initial guess of $v = \frac{dy}{dt}$? 

```python
t_arr = np.arange(t0, tf+2*step, step)
y_arr = np.zeros(len(t_arr))
v_arr = np.zeros(len(t_arr))

z = z2
v0 = 10 #0.01 or 10

r = np.array([y0, v0],float)
for t in t_arr:
    y_arr[i] = r[0]
    v_arr[i] = r[1]
    k1 = step*f(r,t,z)
    k2 = step*f(r+0.5*k1, t+0.5*step, z)
    k3 = step*f(r+0.5*k2, t+0.5*step, z)
    k4 = step*f(r+k3, t+step, z)
    r += (k1 + 2*k2 + 2*k3 + k4)/6

print("The calculated height at t = 1 is", y_arr[-1])

#Plt velocity and height vs. time
plt.scatter(t_arr, y_arr, s=4, label = 'Height')
plt.scatter(t_arr, v_arr, s=4, label = "Velocity")
plt.legend()
plt.xlabel("Time")
plt.grid()
plt.show()
```
