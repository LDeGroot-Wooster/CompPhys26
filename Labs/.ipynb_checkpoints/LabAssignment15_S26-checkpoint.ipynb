### The problem:

Today we will solve a second-order ordinary differential equation for a simple harmonic oscillator. The standard equation is given as: 

$$\frac{d^2x}{dt^2}= -\omega^2 x $$

The general solution to this differential equation is: 

$$x(t) = c_1 \cos \omega t + c_2 \sin \omega t $$

where $c_1$ and $c_2$ can be found from initial conditions. 

For this problem, we will analyze the case of $\omega = 1$ in the range of $t=0$ to $t=50$. The initial conditions are $x=1$ and $dx/dt = 0$. 

*Problem adapted from Exercise 8.6 in the textbook*

### Part 1: Solving the second-order differential equation 

Let's solve for $x$ as a function of time. Do this in the following steps: 

#### Step 1: In a markdown, write the two simultaneous single-order differential equations that you will use to solve this second-order differential equation. 
*Double-check with the professor before moving on.*

$$\frac{dx}{dt} = u$$
$$\frac{du}{dt} = -\omega^2x$$

#### Step 2: Write a function `f(r,t)` that will calculate the two equations written in step 1. 

```python
def f(r,t):
    x = r[0]
    u = r[1]
    #calculates dx/dt and du/dt with above eqns
    f_x = u
    f_u = (-omega**2)*x
    return np.array([f_x,f_u],float) #returns an array
```

#### Step 3: Write a function `RK4_multi(r,t,h,f)` - like you did in the last lab - that you can use to solve any two simultaneous single-order differential equations.  

```python
def RK4_multi(r,t,h,f):
    #first, sets initial x and u to be first values in array
    x = r[0,0]
    u = r[1,0]
    for i in range(len(t)): 
        if i>0: #except for first step, set values of array
            r[0,i] = x
            r[1,i] = u
        #perform RK4
        k_1 = h*f(r_arr[:,i], t_arr[i])
        k_2 = h*f(r_arr[:,i]+0.5*k_1,t_arr[i]+0.5*h)
        k_3 = h*f(r_arr[:,i]+0.5*k_2,t_arr[i]+0.5*h)
        k_4 = h*f(r_arr[:,i]+k_3,t_arr[i]+h)
        #update x and u for next loop
        x += (1/6)*(k_1[0]+2*k_2[0]+2*k_3[0]+k_4[0])
        u += (1/6)*(k_1[1]+2*k_2[1]+2*k_3[1]+k_4[1])
    return r_arr #returns an array
```

#### Step 4: Define any constants and arrays necessary for the problem and solve for $x$ and $dx/dt$ as a function of time. 

```python
#define constants
t_0 = 0 
t_f = 50
num_steps = 1000
h = (t_f-t_0)/num_steps
omega = 1

#define arrays and make first column of array [x,u]
x = 1
u = 0
t_arr = np.arange(t_0,t_f+h,h)
r_arr = np.zeros([2,len(t_arr)],float)
r_arr[0,0] = x
r_arr[1,0] = u
f = f

#compute result
result = RK4_multi(r_arr,t_arr,h,f)
```

### Part 3: Anharmonic Oscilator

$$ \frac{d^2x}{dt} = -\omega^2 x^3 $$

Step 1: Needed equations
$$\frac{dx}{dt} = u$$
$$\frac{du}{dt} = -\omega^2x^3$$

```python
#step 2, function using equations above
def f(r,t):
    x = r[0]
    u = r[1]
    f_x = u
    f_u = (-omega**2)*(x**3)
    return np.array([f_x,f_u],float)

#step 3a initial conditions
t_0 = 0 
t_f = 30
num_steps = 500
h = (t_f-t_0)/num_steps
omega = 1

#step 3b, make array and fill first column
x = 1
u = 0
t_arr = np.arange(t_0,t_f+h,h)
r_arr = np.zeros([2,len(t_arr)],float)
r_arr[0,0] = x
r_arr[1,0] = u
f = f

#compute and plot result
result4 = RK4_multi(r_arr,t_arr,h,f)
plt.scatter(t_arr,result4[0,:], s=4, label='x')
plt.xlabel("Time (seconds)")
plt.ylabel("x (distance units)")
plt.show()
```

#### Step 2: Once again, vary the amplitude to see what happens when we increase the amplitude from $x=1$ to $x=2$ to $x=4$. Plot all three position vs. time graphs on the same graph (be sure to include a legend). What do you notice about an anharmonic oscillator as the amplitude increases? 

```python
#initial conditions
t_0 = 0 
t_f = 30
num_steps = 1000
h = (t_f-t_0)/num_steps
omega = 1

#arrays for x=1
x = 1
u = 0
t_arr = np.arange(t_0,t_f+h,h)
r_arr = np.zeros([2,len(t_arr)],float)
r_arr[0,0] = x
r_arr[1,0] = u
f = f

#x=1 result
result4_2 = RK4_multi(r_arr,t_arr,h,f)

#arrays for x=2
x = 2
u = 0
t_arr = np.arange(t_0,t_f+h,h)
r_arr = np.zeros([2,len(t_arr)],float)
r_arr[0,0] = x
r_arr[1,0] = u
f = f

#x=2 result
result5 = RK4_multi(r_arr,t_arr,h,f)

#arrays for x=4
x = 4
u = 0
t_arr = np.arange(t_0,t_f+h,h)
r_arr = np.zeros([2,len(t_arr)],float)
r_arr[0,0] = x
r_arr[1,0] = u
f = f

#x=4 result
result6 = RK4_multi(r_arr,t_arr,h,f)

#plot results
plt.scatter(t_arr,result4_2[0,:], s=4, label='x=1')
plt.scatter(t_arr,result5[0,:], s=4, label='x=2')
plt.scatter(t_arr,result6[0,:], s=4, label='x=4')
plt.legend()
plt.xlabel("Time (seconds)")
plt.ylabel("x (distance units)")
plt.show()
```

#### Explanation:

Changing the amplitude of the anharmonic oscillator has a different effect than changing the amplitude of the simple harmonic oscillator. Of course the amplitude of the motion still changes, but increasing the amplitude also causes the period to decrease. At amplitude 1, the period is approximately 7-8 s. At amplitude 2, its 3-4 s. At amplitude=4, it's 1-2 seconds.
