## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

```python
#define the function f(x,t) which is dx/dt
def f(x,t):
    return -x**3 + np.sin(t)

#over what range should we solve this? Usually this is defined in the problem

a = 0.0 #the beginning of the range
b = 10.0 #the end of the range

N = 1000 #the number of steps

h = (b-a)/N

x = 0.0 #boundary cond, but will be updated as we go through the loop

tpoints = np.arange(a,b+h,h)
xpoints = np.zeros(len(tpoints))

#use Euler's Method
#take starting x position and add h*f(x,t)

for i in range(len(tpoints)):
    xpoints[i] = x
    x += h*f(xpoints[i],tpoints[i])

plt.scatter(tpoints,xpoints,s=3)

plt.show()


```

Determine the solution fo $N(t)$ given that the number of atoms to start $N_0$ is 1000, and the decay constant $k = 0.05$

## Second-Order Runge-Kutta

Apply the second-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3 + np.sin(t)

a = 0.0 #the beginning of the range
b = 10.0 #the end of the range

N = 100 #the number of steps, Runge-Kutta will require fewer than Euler's

h = (b-a)/N

x = 0.0 #boundary cond, but will be updated as we go through the loop

tpoints = np.arange(a,b+h,h)
xpoints = np.zeros(len(tpoints))

for i in range(len(tpoints)):
    xpoints[i] = x
    k1 = h*f(x,tpoints[i])
    k2 = h*f(x+0.5*k1,tpoints[i]+0.5*h)
    x += k2


plt.scatter(tpoints,xpoints,s=3,c='r')

plt.show()
```

## Fourth-Order Runge-Kutta

Apply the fourth-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3 + np.sin(t)

a = 0.0 #the beginning of the range
b = 10.0 #the end of the range

N = 20 #the number of steps, Runge-Kutta will require fewer than Euler's

h = (b-a)/N

x = 0.0 #boundary cond, but will be updated as we go through the loop

tpoints = np.arange(a,b+h,h)
xpoints = np.zeros(len(tpoints))

for i in range(len(tpoints)):
    xpoints[i] = x
    k1 = h*f(x,tpoints[i])
    k2 = h*f(x+0.5*k1,tpoints[i]+0.5*h)
    k3 = h*f(x+0.5*k2,tpoints[i]+0.5*h)
    k4 = h*f(x+k3,tpoints[i]+h)
    x += 1/6 * (k1+2*k2+2*k3+k4)


plt.scatter(tpoints,xpoints,s=3,c='r')

plt.show()
```
