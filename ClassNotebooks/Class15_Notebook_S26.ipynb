## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

\begin{align*}
    \frac{dx}{dt}&=-x^3+\sin{t}\\
    x(t+h)&\approx x(t)+hf(x,t)\\
    x(t+h)&\approx x + h\cdot\left(-x^3+\sin(t+h)\right)
\end{align*}

```python
def f(x,t):
    return -x**3 + np.sin(t)

a,b = 0.0, 10.0
N = 1000
h = (b-a)/N
x = 0

t_array = np.linspace(a,b,N)
x_array = np.zeros(N)

for i in range(len(x_array)):
    x_array[i] = x
    x += h*f(x,t_array[i])

plt.plot(t_array,x_array)
plt.show()
```

#### Radioactive decay

$$\frac{dN}{dt} = -kN $$

Determine the solution fo $N(t)$ given that the number of atoms to start $N_0$ is 1000, and the decay constant $k = 0.05$

\begin{align*}
    \frac{dN}{N}&=-k~dt\\
    \ln{N}&=C-kt\\
    N(t)&=N_0 e^{-kt}
\end{align*}

```python
def f(N,k,t):
    return -k*N
k=0.05
N0=1000

a,b = 0.0, 10.0
S = 10
h = (b-a)/S
N = N0

t_array = np.linspace(a,b,S)
N_array = np.zeros(S)

for i in range(S):
    N_array[i] = N
    N += h*f(N,k,t_array)

N_actual = N0 * np.exp(-k*t_array)

plt.plot(t_array, N_array, "c*")
plt.plot(t_array, N_actual,"m")
plt.show()
```

## Second-Order Runge-Kutta

Apply the second-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -(x**3) + np.sin(t)

a,b = 0,20
N,h = 100, (b-a)/N
t_array  = np.linspace(a,b,N) 
x_array = np.zeros(N)
x = 0.0

for i in range(N):
    x_array[i] = x
    t = t_array[i]
    k1 = h*f(x,t)
    k2 = h*f(x + k1/2, t + h/2)
    x  += k2
    
plt.plot(t_array,x_array,"*-")
plt.show()
```
