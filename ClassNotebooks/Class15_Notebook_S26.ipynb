## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

```python
# define function f(x,t)
def f(x,t):
    return -x**3 + np.sin(t)

# define initial conditions/constants
a = 0.0 # lower limit, starting time
b = 10.0 # upper limit, end of time interval
N = 1000
h = (b - a)/N
x = 0.0

# define position and time arrays
tpts = np.arange(a, b+h, h)
xpts = np.zeros(len(tpts))

# apply Euler's method
for i in range(len(tpts)):
    xpts[i] = x
    x += h * f(x,tpts[i])

# plot the results
plt.plot(tpts, xpts, ".")
plt.xlabel("time (s)")
plt.ylabel("x(t)")
plt.show()
```

#### Radioactive decay

$$\frac{dN}{dt} = -kN $$

Determine the solution fo $N(t)$ given that the number of atoms to start $N_0$ is 1000, and the decay constant $k = 0.05$

```python
# define function f(N,t,k)
def f(N,t,k):
    return -k*N

# define initial conditions/constants
k = 0.05
N0 = 1000

tstart = 0.0 # lower limit, starting time
tend = 100.0 # upper limit, end of time interval
Nsteps = 1000
h = (tend - tstart)/Nsteps

# define position and time arrays
tpts = np.arange(tstart, tend+h, h)
Npts = np.zeros(len(tpts))

# apply Euler's method
for i in range(len(tpts)):
    Npts[i] = N0
    N0 += h * f(N0,tpts[i],k)

# actual results
Nactual = 1000*np.exp(-k*tpts)

# plot the results
plt.plot(tpts, Npts, ".")
plt.plot(tpts, Nactual, ".")

plt.xlabel("time (s)")
plt.ylabel("N(t)")
plt.show()
```

## Second-Order Runge-Kutta

Apply the second-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
# define function
def f(x,t):
    return -x**3 + np.sin(t)

# define initial conditions/constants
a = 0.0
b = 10.0
N = 100
h = (b-a)/N
x = 0.0

# define position and time arrays
tpts = np.arange(a, b+h, h)
xpts = np.zeros(len(tpts))

# apply second order Runge-Kutta method
for i in range(len(tpts)):
    xpts[i] = x
    k1 = h * f(x, tpts[i])
    k2 = h * f(x+0.5*k1, tpts[i]+0.5*h)
    x += k2

# plot the results
plt.plot(tpts, xpts, ".")
plt.xlabel("time (s)")
plt.ylabel("x(t)")
plt.show()
```

## Fourth-Order Runge-Kutta

Apply the fourth-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
# define function
def f(x,t):
    return -x**3 + np.sin(t)

# define initial conditions/constants
a = 0.0
b = 10.0
N = 30
h = (b-a)/N
# initial conditions
x = 0.0

# define position and time arrays
trk4 = np.arange(a, b+h, h)
xrk4 = np.zeros(len(trk4))

# apply second order Runge-Kutta method
for i in range(len(trk4)):
    xrk4[i] = x
    k1 = h * f(x, trk4[i])
    k2 = h * f(x+0.5*k1, trk4[i]+0.5*h)
    k3 = h * f(x+0.5*k2, trk4[i]+0.5*h)
    k4 = h * f(x+k3, trk4[i]+h)
    x += (k1 + 2*k2 + 2*k3 + k4)/6

# plot the results
plt.plot(trk4, xrk4, ".")
plt.xlabel("time (s)")
plt.ylabel("x(t)")
plt.show()
```
