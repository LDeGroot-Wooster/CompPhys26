## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

```python
def f(x,t):
    return -x**3 + np.sin(t)

a = 0.0
b = 10.0
N = 1000
h = (b-a)/N
x = 0.0

times = np.arange(a, b+h, h)
domain = np.zeros(len(times))

for i in range(len(times)):
    domain[i] = x
    x += h*f(x,times[i])

print(domain)
```

## Fourth-Order Runge-Kutta

Apply the fourth-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3 + np.sin(t)

a = 0.0
b = 10.0
N = 50
h = (b-a)/N
x = 0.0

times = np.arange(a, b+h, h)
domain = np.zeros(len(times))

for i in range(len(times)):
    domain[i] = x
    k1 = h*f(x, times[i])
    k2 = h*f(x+0.5*k1, times[i]+0.5*h)
    k3 = h*f(x+0.5*k2, times[i]+0.5*h)
    k4 = h*f(x+0.5*k3, times[i]+0.5*h)
    x += k4

plt.scatter(times, domain)
plt.show()
```
