## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

```python
def f(x,t):
    return -x**3+np.sin(t)

a = 0.0 
b = 10.0
N = 20
h = (b-a)/N
x_0 = 0.0

t_arr = np.arange(a,b+h,h)
x_arr = np.zeros(len(t_arr))

for i in range(len(t_arr)):
    x_arr[i] = x_0
    x_0 += h*f(x_0, t_arr[i])

plt.scatter(t_arr,x_arr, s=4)
plt.xlabel("Time (seconds)")
```

## Second-Order Runge-Kutta

Apply the second-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3+np.sin(t)

a = 0.0 
b = 10.0
N = 20
h = (b-a)/N
x_0 = 0.0

t_arr = np.arange(a,b+h,h)
x_arr = np.zeros(len(t_arr))

for i in range(len(t_arr)):
    x_arr[i] = x_0
    k_1 = h*f(x_0, t_arr[i])
    k_2 = h*f(x_0+0.5*k_1,t_arr[i]+0.5*h)
    x_0 += k_2

plt.scatter(t_arr,x_arr, s=4)
plt.xlabel("Time (seconds)")
```

## Fourth-Order Runge-Kutta

Apply the fourth-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3+np.sin(t)

a = 0.0 
b = 10.0
N = 20
h = (b-a)/N
x_0 = 0.0

t_arr = np.arange(a,b+h,h)
x_arr = np.zeros(len(t_arr))

for i in range(len(t_arr)):
    x_arr[i] = x_0
    k_1 = h*f(x_0, t_arr[i])
    k_2 = h*f(x_0+0.5*k_1,t_arr[i]+0.5*h)
    k_3 = h*f(x_0+0.5*k_2,t_arr[i]+0.5*h)
    k_4 = h*f(x_0+k_3,t_arr[i]+h)
    x_0 += (1/6)*(k_1+2*k_2+2*k_3+k_4)

plt.scatter(t_arr,x_arr, s=4)
plt.xlabel("Time (seconds)")
```
