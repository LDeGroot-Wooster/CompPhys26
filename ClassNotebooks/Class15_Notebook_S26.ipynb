## Euler's Method  
#### Example 8.1

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 

```python
# define the function f(x,t)
def f(x, t):
    return -x**3 + np.sin(t)

# startig conditions and constants
a = 0.0
b = 10.0 #end time
N = 763 #steps
h = (b-a)/N
x = 0.0

# define position array
tpts = np.arange(a, b+h, h)
xpts = np.zeros(len(tpts))

# eulers
for i in range(len(tpts)):
    xpts[i] = x 
    x += h*f(x,tpts[i])

plt.plot(tpts, xpts, '.')
plt.xlabel('time (s)')
plt.ylabel('x(t)')
```

#### Radioactive decay

$$\frac{dN}{dt} = -kN $$

Determine the solution fo $N(t)$ given that the number of atoms to start $N_0$ is 1000, and the decay constant $k = 0.05$

```python
def f(N, k):
    return -k*N

tstart = 0.0
tend = 2.0 #end time
Num = 763 #steps
h = (tstart-tend)/Num
N_0 = 1000
k = 0.05

# define position array
tpts = np.arange(tstart, tend+h, h)
Npts = np.zeros(len(tpts))

for i in range(len(tpts)):
    Npts[i] = N_0 
    N_0 += h*f(N,tpts[i],k)

plt.plot(tpts, Npts, '.')
plt.xlabel('time (s)')
plt.ylabel('x(t)')
```

## Second-Order Runge-Kutta

Apply the second-order Runge-Kutta method to solve example 8.1 as done above with Euler's method

Solve the differential equation: 
$$ \frac{dx}{dt} = -x^3 +\sin t$$
with initial conditions $x=0$ at $t=0$. 


```python
def f(x,t):
    return -x**3 + np.sin(t)

a = 0.0
b = 10.0
N = 100
h = (b-a)/N

x = 0.0
tpts = np.arange(a, b+h, h)
xpts = np.zeros(len(tpts))

for i in range(len(tpts)):
    xpts[i] = x 
    k1 = h*f(x,tpts[i])
    k2 = h*f(x + 0.5*k1, tpts[i] + 0.5*h)
    x += k2

plt.plot(tpts, xpts, '.')
plt.xlabel('time (s)')
plt.ylabel('x(t)')
```
